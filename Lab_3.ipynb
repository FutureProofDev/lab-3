### Part 1.1 — A single neuron is just a line
A neuron computes `y_hat = w * x + b`. Fit it to the sine data with **manual gradient descent**.

```python
# Toy Data Generation:
x = np.linspace(-3, 3, 100)

gaus_noise = np.random.normal(loc = 0, scale=0.1, size = len(x))
y = np.sin(x) + gaus_noise


plt.plot(x, y)

# Random Weight Initialization

w = np.random.randn() * 0.01
b = np.random.randn() * 0.01



lr = 0.1   # learning rate
loss_log = []

# One Gradient-Descent
for i in range(200):
    y_hat = w * x + b
    loss  = np.mean((y_hat - y) ** 2)               # MSE
    dw    = np.mean(2 * (y_hat - y) * x)            # dLoss/dw  (derive this on paper!)
    db    = np.mean(2 * (y_hat - y))                # dLoss/db
    w    -= lr * dw
    b    -= lr * db
    loss_log.append(loss)


# Plotting
plt.figure(figsize=(10, 4))


plt.subplot(1, 2, 1)
plt.plot(loss_log)
plt.title("Loss over iterations")
plt.ylabel("Loss (MSE)")
plt.xlabel("Iteration")

plt.subplot(1, 2, 2)
plt.scatter(x, y, label="Data")
plt.plot(x, y_hat, color="red", label="Fitted line")
plt.xlabel("x")
plt.ylabel("y")
plt.title("Model fitting")


plt.show()
```

**Student Reasoning — The single neuron**
*1. Derive `dLoss/dw` on paper and include a photo or the LaTeX in this cell — does it match
the code?*
*2. The loss stops improving but the fit is clearly bad. Is this underfitting or overfitting,
and WHY can a single neuron never fit a sine wave, no matter how long you train?*

> **Answer:** 
LaTeX Proof

For a single point $i$, let the Mean Squared Error loss $L$ be:
$L = (\hat{y}_i - y_i)^2$

where $\hat{y}_i = w \cdot x_i + b$.

By the chain rule:
$\frac{dL}{dw} = \frac{dL}{d\hat{y}_i} \cdot \frac{d\hat{y}_i}{dw}$

Differentiating $L$ with respect to $\hat{y}_i$:
$\frac{dL}{d\hat{y}_i} = 2(\hat{y}_i - y_i)$

Differentiating $\hat{y}_i$ with respect to $w$:
$\frac{d\hat{y}_i}{dw} = x_i$

Substituting back:
$\frac{dL}{dw} = 2(\hat{y}_i - y_i)x_i$

For $N$ points, taking the average over all data points yields:
$\frac{dL}{dw} = \frac{1}{N} \sum_{i=1}^{N} 2(\hat{y}_i - y_i)x_i$

1. Yes it matches the code. Where the sum notation and division (1 / N) is represented by the np.mean function. And the expression within the sum notation is represented by (2 * (y_hat - y) * x)

2. This is underfitting. The model is too simple and fails to accurately match the given equation. Therefore, it's output values would be inaccurate.
A single neuron can never fit a sine wave because a single neuron generates a linear equation (line) and a sine wave is not a linear equation but rather a curve.





### Part 1.2 — Hidden layers and activations: why depth helps
Now build a tiny network **by hand**: 1 input -> a hidden layer of 8 neurons with `tanh` ->
1 output. Still NumPy, still manual gradients.

```python
# Paramter Initialization
x = x.reshape(-1, 1)
y = y.reshape(-1, 1)

W1 = np.random.randn(1, 8) * 0.5
W2 = np.random.randn(8, 1) * 0.5

b1 = np.random.randn(8,) * 0.5
b2 = np.random.randn(1,) * 0.5



lr = 0.05  # learning rate
loss_log = []

for i in range(3000):
    # Forward pass
    z1 = x @ W1 + b1        # pre-activation, shape (100, 8)
    h  = np.tanh(z1)        # activation
    y_hat = h @ W2 + b2     # output, shape (100, 1)
    loss  = np.mean((y_hat - y) ** 2)
    hint = h

    # Backward pass 
    d_yhat = 2 * (y_hat - y) / len(y)
    dW2 = h.T @ d_yhat 
    db2 = d_yhat.sum(axis=0)

    dh  = d_yhat @ W2.T
    dz1 = dh * (1 - np.tanh(z1) ** 2)     # derivative of tanh
    dW1 = x.T @ dz1 ;  db1 = dz1.sum(axis=0)   


    W1 -= lr * dW1
    b1 -= lr * db1
    W2 -= lr * dW2
    b2 -= lr * db2
    loss_log.append(loss)


# Plotting 
plt.figure(figsize=(10, 4))

# Loss
plt.subplot(1, 2, 1)
plt.plot(loss_log)
plt.title("Loss over iterations")
plt.ylabel("Loss (MSE)")
plt.xlabel("Iteration")

# Model Fit
plt.subplot(1, 2, 2)
plt.scatter(x, y, label="Data")
plt.plot(x, y_hat, color="red", label="Fitted line")
plt.xlabel("x")
plt.ylabel("y")
plt.title("Model fitting")



# Without tan h

for i in range(3000):
    # Forward pass
    z1 = x @ W1 + b1        # pre-activation, shape (100, 8)
    h  = z1                 # NO activation
    y_hat = h @ W2 + b2     # output, shape (100, 1)
    loss  = np.mean((y_hat - y) ** 2)

    # Backward pass 
    d_yhat = 2 * (y_hat - y) / len(y)
    dW2 = h.T @ d_yhat 
    db2 = d_yhat.sum(axis=0)

    dh  = d_yhat @ W2.T
    dz1 = dh              
    dW1 = x.T @ dz1 ;  db1 = dz1.sum(axis=0)   


    W1 -= lr * dW1
    b1 -= lr * db1
    W2 -= lr * dW2
    b2 -= lr * db2
    loss_log.append(loss)

# Plotting new fit

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.scatter(x, y, label="Data")
plt.plot(x, y_hat, color="red", label="Fitted line")
plt.xlabel("x")
plt.ylabel("y")
plt.title("No Activation Function (tanh) \n Model Fitting")    


plt.subplot(1, 2, 2)
for i in range(4):  
    plt.plot(x, hint[:, i])
plt.xlabel("x")
plt.ylabel("h")
plt.title("Hint - Tanh against x")

plt.show()
```

Taking layer two for example, expanding $\hat{y}$ as input :


$\hat{y}$ = (X $\cdot$ w1 + b1) $\cdot$ w2 + b2

The addition of numerous linear layers would just result in another linear model.


2. Each of neurons contributed a different curve or bend that allowed the linear model to change shape and be able to fit onto the given curve.


3. Backpropagation is a method that uses the chain rule to evaluate how much each weight or bias contributed to the loss generated in an iteration or run. Based on the different losses with respect to weights or biases, parameters are then adjusted to minimize the loss in subsequent runs.

### Part 1.3 — The learning rate: too cold, too hot, just right
The single most important hyperparameter. See its three regimes with your own eyes.

```python
def train_toy(lr, steps = 3000):

    np.random.seed(42)

    X = x.reshape(-1, 1)
    Y = y.reshape(-1, 1)

    W1 = np.random.randn(1, 8) * 0.5
    W2 = np.random.randn(8, 1) * 0.5

    b1 = np.random.randn(8,) * 0.5
    b2 = np.random.randn(1,) * 0.5

    loss_log = []

    for i in range(steps):
        # Forward pass
        z1 = X @ W1 + b1        # pre-activation, shape (100, 8)
        h  = np.tanh(z1)        # activation
        y_hat = h @ W2 + b2     # output, shape (100, 1)
        loss  = np.mean((y_hat - Y) ** 2)
        

        # Backward pass 
        d_yhat = 2 * (y_hat - Y) / len(y)
        dW2 = h.T @ d_yhat 
        db2 = d_yhat.sum(axis=0)

        dh  = d_yhat @ W2.T
        dz1 = dh * (1 - np.tanh(z1) ** 2)     # derivative of tanh
        dW1 = X.T @ dz1 ;  db1 = dz1.sum(axis=0)   


        W1 -= lr * dW1
        b1 -= lr * db1
        W2 -= lr * dW2
        b2 -= lr * db2

        loss_log.append(loss)

    return loss_log


# Different learning rates
loss_curve1 = train_toy(0.001)    
loss_curve2 = train_toy(0.05)    
loss_curve3 = train_toy(1.0)   

# Plotting on the same axes
plt.figure(figsize=(8, 4))
plt.plot(loss_curve1, label="lr = 0.001")
plt.plot(loss_curve2, label="lr = 0.05")
plt.plot(loss_curve3, label="lr = 1.0")

plt.yscale('log')  # Log scale on y-axis per the TODO hint!
plt.ylim(1e-3, 1000)
plt.xlabel("Iteration")
plt.ylabel("Loss ")
plt.title("Learning Rate Comparison")
plt.legend()
plt.show()

```

**Student Reasoning — Learning rate**
*Describe what each of the three curves shows. Why does a too-large learning rate make the
loss explode or oscillate instead of just "learning faster"?*

> **Answer:** 

The 0.001 learning rate curve is very gradual and shows a slow change over time.

The 0.05 shows a steeping curve in the beginning iterations before evening out relatively quickly.

The 1.0 learning root shoots up quickly into a vertical line.


A large learning rate leads to greater adjustments that compound from each run to each run leading to an exponential increase  seen as an explosion. Rather than gradually finding an optimum, it overshoots that right value and continues to overshoot from iteration to iteration resulting in "infinite" loss.
